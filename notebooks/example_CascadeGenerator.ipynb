```python
%load_ext autoreload
%autoreload 2
from random import random
import networkx as nx
import cascadesimulator.cascade_generator_cpp as cg
```

```python
graph = nx.erdos_renyi_graph(100, 0.1)
symptom_rates = [0.5 + 0.5*random() for u in graph.nodes()]
for edge in graph.edges():
    graph[edge[0]][edge[1]]['weight'] = 0.1 + 0.2*random()
cascade_model = cg.CascadeGenerator()
```

```python
adj_list_ = [[v for v in graph.neighbors(u)] for u in graph.nodes()]
cascade_model.set_graph(adj_list_)
probs = [[graph[u][v]['weight'] for v in graph.neighbors(u)] for u in graph.nodes()]
cascade_model.set_probabilities(probs)
if symptom_rates is not None:
    cascade_model.set_symptom_probabilities(symptom_rates)
```

```python
seeds = [0]

cascade_model.generate_cascade(seeds)
```

```python
num_samples = 10

cascade_model.generate_cascades(seeds, num_samples)
```
